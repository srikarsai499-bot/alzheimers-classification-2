## 33. Correlation Heatmap

```python
# PatientID and DoctorInCharge are excluded because they are not useful
# for numerical correlation analysis.

eda_df = df.drop(columns=["PatientID", "DoctorInCharge"])

correlation = eda_df.corr()

plt.figure(figsize=(18, 14))

sns.heatmap(
    correlation,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")

plt.show()
```

## 34. Feature Correlation with Diagnosis

```python
correlation_with_target = (
    eda_df.corr()["Diagnosis"]
    .sort_values(ascending=False)
)

print(correlation_with_target)
```

```python
plt.figure(figsize=(8, 10))

correlation_with_target.drop("Diagnosis").sort_values().plot(
    kind="barh"
)

plt.title("Feature Correlation with Alzheimer's Diagnosis")
plt.xlabel("Correlation")
plt.ylabel("Feature")

plt.show()
```
